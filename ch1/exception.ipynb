```python
# try ~ except

names = ["kim","park","Lee"]

try:
    z = "Let"
    # 예외 발생 가능성 구문
    x = names.index(z)
    print(f"{z} Found it!")
except:
    print("에러 발생")
```

```python
# try ~ except ~ else

try:
    age = int(input("나이 입력"))
except:
    print("입력이 정확하지 않습니다.")
else: # 오류가 없다면
    if age <= 18:
        print("미성년자는 출입금지 입니다.")
    else:
        print("환영합니다.")
```

```python
# 오류 강제 발생 : raise 오류명

class Bird:
    def fly(self):
        raise NotImplementedError

class Eagle(Bird):
    def fly(self):
        print("very fast")

eagle = Eagle()
eagle.fly()
```

```python
# 예외 클래스 생성 가능
class MyError(Exception):
    def __str__(self):
        return "허용되지 않는 별명입니다."
    
def say_nick(nick):
    if nick == "바보":
        raise MyError()
    print(nick)

try:
    say_nick("천사")
    say_nick("바보")
except MyError as e:
    print(e)
```

```python
# 문제 1. 랜덤 비밀번호 생성기

# 사용자로부터 생성할 비밀번호의 길이를 입력받아 랜덤 비밀번호를 생성하는 프로그램 작성
chars = "ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz0123456789!@#$%"
# 비밀번호 길이는 4 이상만 허용
# 숫자가 아닌 값을 입력하면 예외 처리한다

import random

while True:
    try:
        pwd_num = int(input("비밀번호 길이 : "))

        if pwd_num < 4:
            print("비밀번호는 4자리 이상이어야 합니다.")
            continue
        break         
    except ValueError:
        print("숫자만 입력하세요")
# 리스트 컴프리헨션
# password = []
# for i in range(pwd_num):
#     password.append(random.choice(chars))
password = [random.choice(chars) for _ in range(pwd_num)]

# print(f"랜덤 비밀번호 : {password}")
# 리스트 => 문자열로 변경
print(f"랜덤 비밀번호 : {"".join(password)}")
```

```python
# 나이 계산 프로그램
# 생년월일을 입력받아 현재 나이를 계산하는 프로그램 작성
# 입력형식 2001-08-25
# input() => '2001-08-25' => 날짜 형식

import datetime

# 오늘날짜
now = datetime.datetime.now()

try:
    # 생년월일 입력
    birth_str = input("생년월일 입력(YYYY-MM-DD) ")
    # '2001-05-15'
    birth = datetime.datetime.strptime(birth_str,"%Y-%m-%d")

    # 2026 - 2001
    # now 년도만 추출
    age = now.year - birth.year
    if (now.month, now.day) < (birth.month, birth.day):
        age -= 1

    print("나이 ", age)
except:
    print("날짜 형식이 올바르지 않습니다.")
```

```python
# 다음과 같은 csv 파일이 있다.
# 이름,국어,영어,수학
# 홍길동,90,80,70
# 홍일동,95,50,40
# 김길동,55,20,70
# 감길동,20,10,40

# csv 파일을 읽어 학생별 평균 점수 출력
# 홍길동 : 80.0

# csv.DictReader() 사용, 평균은 둘째자리까지
import csv
list1 = [
    {"이름" : "홍길동","국어" : 90, "영어" : 80, "수학" : 70},
    {"이름" : "홍일동","국어" : 95, "영어" : 50, "수학" : 40},
    {"이름" : "김길동","국어" : 55, "영어" : 20, "수학" : 70},
    {"이름" : "감길동","국어" : 20, "영어" : 10, "수학" : 40},
]

with open("./data/stu1.csv","w",encoding="utf-8",newline="") as f:
    headers = ['이름', '국어', '영어', '수학']
    writer = csv.DictWriter(f, fieldnames=headers)
    writer.writeheader()
    for row in list1:
        writer.writerow(row)

with open("./data/stu1.csv","r",encoding="utf-8") as f:
    reader = csv.DictReader(f)
    # print(reader)
    # print(row.get('이름'))
    for c in reader:
        # print(c.get('이름'))
        avg = (int(c.get('국어')) + int(c.get('영어')) + int(c.get('수학'))) /3
        print(f"{c.get('이름')} : {avg:.2f}")





   
```
