```python
from faker import Faker

fake = Faker(locale="ko_KR")
Faker.seed(4321)

print(fake.name())
print(fake.email())
print(fake.address())
print(fake.ssn())
print(fake.phone_number())
```

```python
from faker.providers import internet

fake.add_provider(internet)

for i in range(10):
    print(fake.ipv4_private())
```

```python
fake.date_of_birth(minimum_age=20, maximum_age=30)
fake.date_of_birth(minimum_age=20, maximum_age=30).strftime("%Y-%m-%d")
```

```python
fake.date_time_between(start_date="-3y",end_date="now")
```

```python
# maru_member 50명의 임의의 데이터를 이용해 삽입
# name,birth,phone,joined_at

# 50명의 데이터를 리스트에 추가


# insert

# sql = "insert into maru_member"


data = []

# name
for _ in range(50):
        name = fake.name()
        birth = fake.date_of_birth(minimum_age=20, maximum_age=70).strftime("%Y-%m-%d")
        phone = fake.phone_number()
        joined_at = fake.date_time_between(start_date="-2y",end_date="now")
        data.append([name,birth,phone,joined_at])

print(data)


# insert

sql = "insert into maru_member(name,birth,phone,joined_at) values(:1, :2, :3, :4)"


with oracledb.connect(user="python_user",password=password,dsn="localhost/xe") as conn:
    with conn.cursor() as cursor:
        cursor.executemany(sql,data)
        conn.commit()
    
```
